```python
!pip install ultralytics
import os
from ultralytics import YOLO
dataset_path = "/kaggle/input/datasets/pranshutripathi29/yolov26/dataset"
dataset_name = os.listdir(dataset_path)[0]

data_path = f"{dataset_path}/{dataset_name}"

def main():
    # Load pretrained YOLOv11 nano model
    model = YOLO("yolo11n.pt")   # lightweight & fast

    # Train the model
    model.train(
        data="/kaggle/input/datasets/pranshutripathi29/yolov26/dataset/data.yaml",   # path to your YAML
        epochs=50,             # you can increase to 100 for better accuracy
        imgsz=640,             # matches your dataset preprocessing
        batch=16,              # reduce to 8 if GPU/PC is weak
        name="equation_model", # folder name in runs/detect/
        device="cpu"           # change to 0 if using GPU
    )

if __name__ == "__main__":
    main()
```
